### Question 1:

Given a list of strings, count how many times each string appears in the list. Store the results in a dictionary named `your_ans`, where each string is a key and its frequency is the corresponding value.

For eg: 
```python
animals = ['dog', 'cat', 'elephant', 'cat', 'dog', 'dog', 'dog', 'fox', 'elephant']
```

Your answer should be: 

```
{ 'dog': 4, 'cat': 2, 'elephant': 2, 'fox': 1 }
```

````python
your_ans = {}

animals = ['dog', 'cat', 'elephant', 'cat', 'dog', 'dog', 'dog', 'fox', 'elephant']

for animal in animals:
    if(animal not in your_ans):
        your_ans[animal] = 1
    else:
        your_ans[animal] += 1


your_ans
````

### Question 2:

Suppose a passing score for a course is 60 out of 100. The variable `scores` contains the exam scores of all students in the course.

Write a Python program that uses only **one for loop** to determine:
- the highest score,
- the lowest score, and
- the total number of students who passed the exam.

For eg:

```python
scores = [55, 59, 71, 62, 40, 90, 12, 85, 58, 22, 91]
```

Your output should be: 

```python
Highest: 91 Lowest: 12
Total Passed: 5
```

Restriction: **Do not use** python in-built functions `max()`, `min()`, `sum()`

````python
scores = [55, 59, 71, 62, 40, 90, 12, 85, 58, 22, 91]

highest  = 0
lowest = 100
total_passed = 0

for score in scores:
    if(score > highest):
        highest = score

    if(score < lowest):
        lowest = score

    if(score >= 60):
        total_passed += 1

print(f"Highest: {highest} Lowest: {lowest}")
print(f"Total Passed: {total_passed}")
````

### Question 8:

Given a list of person's name and age as shown below, find the name of the person with the youngest age. Assume no two people have same name or age.

For eg:

```python
persons = [('Kevin', 23), ('Sarah', 24), ('Justin', 21), ('Peter', 24), ('Chris', 32)]
```

The youngest person is `Chris`

````python
persons = [('Kevin', 23), ('Sarah', 24), ('Justin', 21), ('Peter', 24), ('Chris', 32)]
````

````python
youngest = persons[0][0]
min_age = persons[0][1]

for name, age in persons:
    if(age < min_age):
        min_age = age
        youngest = name

print(f"The youngest person is: {youngest}")
````

### Question 9:

Write a Python program to calculate the sum of the elements row-wise of a matrix of `nrows` rows and `ncols` columns. Store your answer in the variable `your_ans`.


For example:
```python
matrix = [
    [4, 2, 7, 1, 1],
    [1, 5, 3, 2, 5],
    [8, 6, 9, 1, 0]
]
```
The column-wise sum of elements are:
```text
4+2+7+1+1 = 15
1+5+3+2+5 = 16
8+6+9+1+0 = 24
```
Your answer should be `[15, 16, 24]`

Restriction: Do not use python in-built `sum` function

_Hint_: Use nested loops

````python
matrix = [
    [4, 2, 7, 1, 1],
    [1, 5, 3, 2, 5],
    [8, 6, 9, 1, 0]
]
````

````python
nrows = len(matrix) # total number of rows
ncols = len(matrix[0]) # total number of cols

your_ans = []

for r in range(nrows):
    total = 0
    for c in range(ncols):
        total += matrix[r][c]

    your_ans.append(total)


your_ans
````

### Question 14:

Write a python code to compute the following sum:

$$
\frac{1}{\sqrt{1} + \sqrt{2}} + \frac{1}{\sqrt{2} + \sqrt{3}} + \frac{1}{\sqrt{3} + \sqrt{4}} + \cdots + \frac{1}{\sqrt{624} + \sqrt{625}}
$$

Store your answer in the variable `your_ans` rounded upto two decimal points.

````python
your_ans = 0

for i in range(1, 625):
    your_ans += 1 / (math.sqrt(i) + math.sqrt(i+1))

your_ans = round(your_ans, 2)
your_ans
````
