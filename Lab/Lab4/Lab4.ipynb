````python
import numpy as np
import plotly.graph_objects as go

bins = np.array([0, 2, 4, 12, 18, 36])

counts, _ = np.histogram(nba['Salary'], bins=bins)

fig = go.Figure()

fig.add_trace(go.Bar(
    x=bins[:-1],
    y=counts,
    width=np.diff(bins),
    offset=0,
    marker=dict(
        color='steelblue',
        line=dict(color='black', width=1)
    )
))

fig.update_layout(
    xaxis_title="Range of Salary (in million dollars)",
    yaxis_title="Number of Players",
    bargap=0,
    plot_bgcolor='white'
)

fig.update_xaxes(
    tickmode='array',
    tickvals=bins,
    range=[0, 36],
    showline=True,
    linecolor='black'
)

fig.update_yaxes(
    showline=True,
    linecolor='black'
)

fig.show()
````
