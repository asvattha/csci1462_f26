# 1. What Is a Function?

A **function** is a reusable piece of code that performs a specific task.

A function is called by writing its name followed by parentheses:

```python
function_name(...)
```

Some functions accept one or more **arguments** inside the parentheses.

For example:

```python
print("Hello")
```

Here:

- `print` is the function name.
- `"Hello"` is an argument passed to the function.

# 2. Python Built-in Functions

Python includes many functions that are available automatically.

You do **not** need to import a module before using a built-in function.

Examples include:

```python
print()
type()
int()
float()
round()
len()
abs()
```

In this notebook, we will focus on the first five.

## 2.1 `print()`

The `print()` function displays information on the screen.

It can print:

- Text
- Numbers
- Variables
- Results of expressions
- Multiple values at once

````python
print("Welcome to Data Analytics!")
print(42)

course = "CSCI 1462"
students = 25

print(course)
print("Number of students:", students)
````

### Printing multiple values

Separate values with commas inside `print()`.

Python automatically inserts a space between them.

````python
name = "Alex"
score = 93

print("Student:", name, "Score:", score)
````

### Printing expressions

The expression is evaluated first, and then the result is printed.

````python
x = 8
y = 3

print(x + y)
print(x * y)
print(x / y)
````

## 2.2 `type()`

The `type()` function tells you the data type of a value or variable.

This is especially useful when debugging data analytics code because many errors occur when a value has a different type than you expected.

````python
age = 20
price = 19.99
name = "Taylor"
is_active = True

print(type(age))
print(type(price))
print(type(name))
print(type(is_active))
````

Typical output types include:

| Python type | Meaning | Example |
|---|---|---|
| `int` | Integer | `25` |
| `float` | Decimal / floating-point number | `3.14` |
| `str` | String / text | `"hello"` |
| `bool` | Boolean value | `True`, `False` |

A useful debugging pattern is:

```python
print(value)
print(type(value))
```

This lets you inspect both the value and its type.

## 2.3 `int()`

The `int()` function converts a compatible value to an integer.

Examples:

```python
int("25")
int(8.9)
```

Be careful: converting a floating-point value with `int()` **does not round**. It removes the fractional part by truncating toward zero.

````python
a = int("25")
b = int(8.9)
c = int(-8.9)

print(a)
print(b)
print(c)
````

## 2.4 `float()` — Python's Floating-Point Type

Python does **not** have a built-in function named `double()`.

Instead, Python uses:

```python
float()
```

for floating-point numbers.

On typical Python implementations, a Python `float` uses **double-precision floating-point representation** internally.

So, for introductory Python:

- Use `int()` for integers.
- Use `float()` for decimal / floating-point values.
- Do not use `double()` as a Python built-in function.

````python
x = float("3.14")
y = float(5)

print(x)
print(y)

print(type(x))
print(type(y))
````

### A common data analytics situation

Data read from a file or user input may initially be stored as text.

For example:

```python
price = "12.75"
```

Before doing arithmetic, you may need to convert it:

```python
price = float(price)
```

````python
price_text = "12.75"
quantity = 4

price = float(price_text)
total = price * quantity

print("Total:", total)
````

## 2.5 `round()`

The `round()` function rounds a number.

Basic form:

```python
round(number)
```

You can also specify the number of decimal places:

```python
round(number, digits)
```

````python
value = 12.6789

print(round(value))
print(round(value, 1))
print(round(value, 2))
print(round(value, 3))
````

### Why `round()` is useful in data analytics

Calculations involving averages, percentages, rates, and measurements often produce many decimal places.

For presentation, you may want to round the result:

```python
average = 87.43627
print(round(average, 2))
```

This displays:

```text
87.44
```

The unrounded value can still be retained separately if needed for further calculations.

## 2.6 `len()`

The `len()` function returns the **number of items** in an object. It is commonly used with strings, lists, tuples, dictionaries, and other collections.

For a string, `len()` counts characters. For a list, it counts the number of items.

````python
course = "Data Analytics"
scores = [88, 92, 79, 95]

print(len(course))
print(len(scores))
````

### Data analytics example

If a list contains observations from a dataset, `len()` tells you how many observations are present.

````python
temperatures = [81, 84, 86, 83, 88]

print("Number of observations:", len(temperatures))
````

## 2.7 `str()`

The `str()` function converts a value into a **string**. This is useful when preparing numbers for display or combining values with text.

````python
age = 20
price = 14.75

age_text = str(age)
price_text = str(price)

print(age_text, type(age_text))
print(price_text, type(price_text))
````

A common beginner mistake is trying to concatenate text and a number directly:

```python
age = 20
message = "Age: " + age
```

That causes a `TypeError`.

One fix is:

```python
message = "Age: " + str(age)
```

For simple output, this is often even easier:

```python
print("Age:", age)
```

## 2.8 `abs()`

The `abs()` function returns the **absolute value** of a number. Absolute value measures distance from zero, so the result is nonnegative.

````python
print(abs(12))
print(abs(-12))
print(abs(-3.75))
````

### Data analytics example

Absolute value is useful for measuring how far a prediction is from an actual value.

````python
actual = 82
predicted = 79

absolute_error = abs(actual - predicted)
print("Absolute error:", absolute_error)
````

## 2.9 `max()`

The `max()` function returns the **largest value** among the values you provide. It can accept individual values or a collection such as a list.

````python
print(max(5, 12, 3, 9))

scores = [88, 72, 95, 81, 90]
print(max(scores))
````

## 2.10 `min()`

The `min()` function returns the **smallest value** among the values you provide.

````python
print(min(5, 12, 3, 9))

scores = [88, 72, 95, 81, 90]
print(min(scores))
````

### Using `min()` and `max()` together

These functions are especially useful when quickly exploring numerical observations.

````python
temperatures = [81, 84, 79, 92, 88, 85]

print("Lowest temperature:", min(temperatures))
print("Highest temperature:", max(temperatures))
````

## 2.11 Built-in `pow()`

Python provides a **built-in** function named `pow()`:

```python
pow(base, exponent)
```

For example, `pow(2, 3)` calculates \(2^3\).

Unlike `math.pow()`, built-in `pow()` does **not** require importing the `math` module.

````python
print(pow(2, 3))
print(pow(5, 2))
print(type(pow(5, 2)))
````

### Built-in `pow()` vs. `math.pow()` vs. `**`

All three can perform exponentiation:

```python
5 ** 2
pow(5, 2)
math.pow(5, 2)
```

A useful beginner distinction is:

- `5 ** 2` → `25`
- `pow(5, 2)` → `25`
- `math.pow(5, 2)` → `25.0`

`math.pow()` converts its arguments to floating-point values and returns a `float`. For basic exponentiation, `**` or built-in `pow()` is usually simpler.

````python
import math

a = 5 ** 2
b = pow(5, 2)
c = math.pow(5, 2)

print(a, type(a))
print(b, type(b))
print(c, type(c))
````

# 3. What Is a Module?

A **module** is a file or collection of Python code containing useful functions, constants, classes, and other definitions.

Modules help organize functionality so that Python does not need to make every possible feature available automatically.

Python includes many modules in its **standard library**, such as:

- `math`
- `random`
- `statistics`
- `datetime`

Other libraries used in data analytics, such as `pandas`, `numpy`, and `matplotlib`, are installed separately.

Before using items from a module, you usually need to **import** it.

````python
import math

print("The math module has now been imported.")
````

# 4. The `math` Module

The `math` module contains useful mathematical constants and functions.

After importing it:

```python
import math
```

you access its contents using **dot notation**:

```python
math.something
```

For example:

```python
math.pi
math.sqrt(25)
```

## 4.1 Mathematical Constants

A **constant** is a named value intended to represent a fixed mathematical quantity.

Two useful constants are:

```python
math.pi
math.e
```

- `math.pi` represents π.
- `math.e` represents Euler's number *e*.

````python
import math

print("pi =", math.pi)
print("e  =", math.e)
````

### Example: area of a circle

The area of a circle is:

$
A = \pi r^2
$

Instead of typing an approximation such as `3.14`, we can use `math.pi`.

````python
import math

radius = 5
area = math.pi * radius ** 2

print("Area:", area)
print("Rounded area:", round(area, 2))
````

## 4.2 `math.sqrt()`

`math.sqrt(x)` returns the square root of `x`.

````python
import math

print(math.sqrt(25))
print(math.sqrt(81))
print(math.sqrt(2))
````

For real-number calculations, `math.sqrt()` expects a nonnegative value.

For example, attempting:

```python
math.sqrt(-4)
```

raises a `ValueError`.

## 4.3 `math.pow()`

`math.pow(x, y)` calculates:

\[
x^y
\]

For example:

```python
math.pow(2, 3)
```

calculates \(2^3 = 8\).

One important detail is that `math.pow()` returns a floating-point value.

````python
import math

print(math.pow(2, 3))
print(math.pow(5, 2))
print(type(math.pow(5, 2)))
````

### `math.pow()` vs. the `**` operator

Python can also perform exponentiation using `**`.

```python
5 ** 2
```

and:

```python
math.pow(5, 2)
```

both represent \(5^2\), but they are not identical in every detail.

For introductory work:

```python
5 ** 2
```

is usually simpler when you only need exponentiation.

````python
import math

print(5 ** 2)
print(math.pow(5, 2))
````

# 5. Built-in Functions vs. Module Functions vs. Methods

This distinction is important.

## Built-in function

A built-in function is available directly:

```python
print("Hello")
type(5)
round(3.14159, 2)
```

There is **no module name before the function**.

---

## Module function

A module function belongs to a module.

After:

```python
import math
```

you call:

```python
math.sqrt(25)
math.pow(2, 3)
```

The module name appears before the function, separated by a dot.

---

## Object method

A **method** belongs to an object.

For example, strings have methods:

```python
name = "python"
name.upper()
```

Here, `upper()` is a method belonging to the string object stored in `name`.

So technically:

- `print()` is a **built-in function**.
- `math.sqrt()` is a **function in the `math` module**.
- `"hello".upper()` is an **object method**.

The words *function* and *method* are sometimes used loosely in conversation, but this distinction is useful when learning Python.

````python
import math

# Built-in function
print(type(10))

# Function from the math module
print(math.sqrt(16))

# Method belonging to a string object
message = "data analytics"
print(message.upper())
````

# 6. Alternative Import Styles

The clearest style for beginners is usually:

```python
import math
```

and then:

```python
math.sqrt(25)
math.pi
```

However, Python also allows:

````python
from math import sqrt, pi

print(sqrt(25))
print(pi)
````

With:

```python
from math import sqrt
```

you can call:

```python
sqrt(25)
```

instead of:

```python
math.sqrt(25)
```

For beginning programmers, using:

```python
import math
```

is often easier to read because it makes the source of the function obvious.

Compare:

```python
sqrt(25)
```

with:

```python
math.sqrt(25)
```

The second version immediately tells the reader that `sqrt` comes from the `math` module.

# 7. Common Mistakes

### Mistake 1: Calling `double()`

```python
double("3.5")
```

Python does not provide a built-in `double()` function.

Use:

```python
float("3.5")
```

---

### Mistake 2: Forgetting to import `math`

```python
math.sqrt(25)
```

will cause a `NameError` if `math` has not been imported in the current kernel session.

Fix:

```python
import math
```

---

### Mistake 3: Forgetting the module name

After:

```python
import math
```

this is incorrect:

```python
sqrt(25)
```

unless `sqrt` was imported separately.

Use:

```python
math.sqrt(25)
```

---

### Mistake 4: Calling a constant like a function

This is incorrect:

```python
math.pi()
```

because `math.pi` is a value, not a function.

Correct:

```python
math.pi
```

---

### Mistake 5: Confusing `int()` with `round()`

```python
int(8.9)
```

produces:

```
8
```

while:

```python
round(8.9)
```

produces:

```
9
```

# 8. Data Analytics Example

Suppose a dataset gives a radius as text:

```python
radius_text = "7.5"
```

We want to:

1. Convert it to a numeric value.
2. Calculate the area of a circle.
3. Round the answer to two decimal places.
4. Display the result.

````python
import math

radius_text = "7.5"

radius = float(radius_text)
area = math.pi * math.pow(radius, 2)
rounded_area = round(area, 2)

print("Radius:", radius)
print("Area:", rounded_area)
print("Area data type:", type(rounded_area))
````

Notice how several ideas work together:

```python
float(radius_text)
```

uses a **built-in function**.

```python
math.pi
```

accesses a **constant from a module**.

```python
math.pow(radius, 2)
```

calls a **function from a module**.

```python
round(area, 2)
```

uses another **built-in function**.

```python
print(...)
```

displays the final result.
