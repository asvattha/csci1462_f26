 <strong style="color:#512888;font-size:18px;">01 · What is a function?</strong><br><span style="color:#38294e;font-size:13px;">Think of a function as a reusable input → processing → output machine.</span></div>

A **function** is a named block of code that performs a specific task. The code inside a function runs **when the function is called**, not when Python first reads its definition.

Think of a coffee machine: you provide ingredients (inputs), it performs steps (processing), and it may produce coffee (output). Similarly, a function may accept **arguments**, perform calculations, and optionally **return** a result.

Functions make programs easier to read, test, reuse, and maintain.

<div style="border-left:6px solid #512888;background:#f5f0fb;border-radius:7px;padding:12px 17px;margin-top:22px;">
 <strong style="color:#512888;font-size:18px;">02 · Built-in functions</strong><br><span style="color:#38294e;font-size:13px;">Python provides many ready-to-use functions without importing a module.</span></div>

````python
print(abs(-12))               # Absolute value → 12
print(round(3.9))              # Nearest integer → 4
print(max(2, 2 + 3, 4))       # Maximum → 5
print(min(8, 3, 12))          # Minimum → 3
print(len("Lumberjacks"))      # String length → 11
print(type(3.14))             # Type → <class 'float'>
````

The function **name** identifies what to do, and the parentheses `()` contain the supplied arguments. Some functions return a value that can be assigned to a variable.

````python
largest = max(14, 9, 27)
print("Largest value:", largest)
````

<div style="border:1px solid #dccbeb;background:#faf7fd;border-radius:8px;padding:12px 16px;"><strong style="color:#512888">Important detail</strong><br>`round()` uses **ties-to-even** behavior in Python: for example, `round(2.5)` is `2`, while `round(3.5)` is `4`. It is not always “round .5 upward.”</div>

<div style="border-left:6px solid #512888;background:#f5f0fb;border-radius:7px;padding:12px 17px;margin-top:22px;">
 <strong style="color:#512888;font-size:18px;">03 · Functions in modules</strong><br><span style="color:#38294e;font-size:13px;">Import a module before calling functions stored inside it.</span></div>

A **module** is a Python file containing reusable definitions. The standard-library `math` module includes square roots, logarithms, and mathematical constants. Use `module.function(...)` to call a function from that module.

A module is not the same thing as a package: a **package** organizes modules (and possibly other packages).

````python
import math

print("Square root of 81:", math.sqrt(81))
print("Log base 2 of 16:", math.log(16, 2))
print("Natural log of e:", math.log(math.e))
print("Pi:", math.pi)
````

````python
distance = math.sqrt((8 - 5)**2 + (1 - 9)**2)
print("Distance:", round(distance, 2))
````

<div style="border-left:6px solid #512888;background:#f5f0fb;border-radius:7px;padding:12px 17px;margin-top:22px;">
 <strong style="color:#512888;font-size:18px;">04 · Defining a function with def</strong><br><span style="color:#38294e;font-size:13px;">Create reusable behavior instead of repeating the same calculation.</span></div>

A function definition begins with **`def`**, followed by a name, parentheses, and a colon. Its body is **indented**. The `return` statement sends a value back to the caller.

```python
def function_name(parameter):
    # indented body
    return result
```

Defining the function does not execute the body; you must **call** the function.

````python
def double(x):
    result = x * 2
    return result

print(double(5))
print(double(12))
````

A returned value can be saved and used later, just like any other Python value.

````python
answer = double(7)
new_total = answer + 3
print("Returned value:", answer)
print("After adding 3:", new_total)
````

<div style="border-left:6px solid #512888;background:#f5f0fb;border-radius:7px;padding:12px 17px;margin-top:22px;">
 <strong style="color:#512888;font-size:18px;">05 · Parameters and arguments</strong><br><span style="color:#38294e;font-size:13px;">Learn the vocabulary used to describe inputs to functions.</span></div>

A **parameter** is a variable named in the function **definition**. An **argument** is the actual value supplied in the function **call**.

In the example below, `item_name`, `quantity`, and `price_per_item` are parameters. The values `"Notebook"`, `3`, and `2.5` are arguments.

````python
def calculate_bill(item_name, quantity, price_per_item):
    total = quantity * price_per_item
    return total

bill = calculate_bill("Notebook", 3, 2.5)
print("Bill amount: $", round(bill, 2), sep="")
````

With **positional arguments**, argument order matters because values are matched to parameters from left to right. Unless you define defaults or variable-length parameters, required parameters must receive arguments.

<div style="border-left:6px solid #512888;background:#f5f0fb;border-radius:7px;padding:12px 17px;margin-top:22px;">
 <strong style="color:#512888;font-size:18px;">06 · Four common function patterns</strong><br><span style="color:#38294e;font-size:13px;">Organize functions by whether they accept parameters and whether they return values.</span></div>

| Function category | Accepts parameters? | Returns a useful value? | Typical use |
|:---|:---:|:---:|:---|
| No parameters, no explicit return | No | No (`None`) | Display a fixed message |
| No parameters, returns a value | No | Yes | Provide a fixed constant or computed result |
| Parameters, no explicit return | Yes | No (`None`) | Display formatted information |
| Parameters, returns a value | Yes | Yes | Compute a result for further use |

In Python, every function returns something. A function that does not explicitly `return` a value returns **`None`**.

#### Pattern A — No parameters, no explicit return

````python
def welcome():
    print("Welcome to CSCI 1462!")

welcome()
````

#### Pattern B — No parameters, returns a value

````python
def get_course_code():
    return "CSCI 1462"

course = get_course_code()
print(course)
````

#### Pattern C — Parameters, no explicit return

````python
def display_student(name):
    print("Student:", name.upper())

display_student("Alex")
````

#### Pattern D — Parameters, returns a value

````python
def calculate_future_value(starting_amount, annual_rate, years):
    future_value = starting_amount * (1 + annual_rate) ** years
    return round(future_value, 2)

forecast = calculate_future_value(100, 0.05, 10)
print("Future value:", forecast)
````

<div style="border-left:6px solid #512888;background:#f5f0fb;border-radius:7px;padding:12px 17px;margin-top:22px;">
 <strong style="color:#512888;font-size:18px;">07 · print() versus return</strong><br><span style="color:#38294e;font-size:13px;">Distinguish displaying a value from passing it back to the caller.</span></div>

`print()` **displays** something in the notebook. `return` **provides** a value to the calling code and immediately ends that function call. A function that only prints does not provide that value for reuse.

````python
def show_double(x):
    print(x * 2)       # Displays the result; implicitly returns None

def get_double(x):
    return x * 2       # Returns the result

printed_result = show_double(5)
returned_result = get_double(5)
print("Value from show_double:", printed_result)
print("Value from get_double:", returned_result)
print("Returned value + 3:", returned_result + 3)
````

<div style="border:1px solid #dccbeb;background:#faf7fd;border-radius:8px;padding:12px 16px;"><strong style="color:#512888">Key takeaway</strong><br>Choose <code>return</code> when another part of the program needs the result. Choose <code>print()</code> when you only want to show information. A function may use both.</div>

<div style="border-left:6px solid #512888;background:#f5f0fb;border-radius:7px;padding:12px 17px;margin-top:22px;">
 <strong style="color:#512888;font-size:18px;">08 · Functions calling other functions</strong><br><span style="color:#38294e;font-size:13px;">Build solutions by combining small reusable pieces.</span></div>

````python
def euclidean_distance(x1, y1, x2, y2):
    return math.sqrt((x1 - x2)**2 + (y1 - y2)**2)

def closer_distance(point_a, point_b, reference_point):
    ax, ay = point_a
    bx, by = point_b
    cx, cy = reference_point
    distance_a = euclidean_distance(ax, ay, cx, cy)
    distance_b = euclidean_distance(bx, by, cx, cy)
    return min(distance_a, distance_b)

print("Distance:", round(euclidean_distance(5, 9, 8, 1), 2))
print("Nearest distance:", round(closer_distance((1, 2), (4, 6), (3, 3)), 2))
````

Notice that `closer_distance()` **reuses** `euclidean_distance()` rather than repeating the distance formula. This is an example of *function composition*.

<div style="border-left:6px solid #512888;background:#f5f0fb;border-radius:7px;padding:12px 17px;margin-top:22px;">
 <strong style="color:#512888;font-size:18px;">09 · Positional and keyword arguments</strong><br><span style="color:#38294e;font-size:13px;">Supply values by position or identify parameters by name.</span></div>

**Positional arguments** are matched by order. **Keyword arguments** use the `parameter=value` format and can be written in a different order. These are ordinary keyword arguments; the special Python notation `**kwargs` refers to accepting an arbitrary set of named arguments and is a separate, more advanced topic.

````python
def item_total(item_name, quantity, price_per_item):
    print(f"{item_name}: ${quantity * price_per_item:.2f}")

item_total("Diet Coke", 5, 1.20)  # Positional arguments
item_total(item_name="Diet Coke", quantity=5, price_per_item=1.20)
item_total(price_per_item=1.20, item_name="Diet Coke", quantity=5)
````

<div style="border:1px solid #dccbeb;background:#faf7fd;border-radius:8px;padding:12px 16px;"><strong style="color:#512888">Argument rule</strong><br>Positional arguments normally come **before** keyword arguments in a call. You cannot assign the same parameter twice (once positionally and again by keyword).</div>

<div style="border-left:6px solid #512888;background:#f5f0fb;border-radius:7px;padding:12px 17px;margin-top:22px;">
 <strong style="color:#512888;font-size:18px;">10 · Default parameter values</strong><br><span style="color:#38294e;font-size:13px;">Make selected arguments optional by declaring defaults.</span></div>

A parameter can have a default value in its definition. If the caller omits that argument, Python uses the default; if the caller provides a value, that value overrides the default.

**Syntax rule:** in ordinary parameter lists, required parameters must come before parameters with defaults.

````python
def calculate_future_value(starting_amount=100, annual_rate=0.05, years=10):
    future_value = starting_amount * (1 + annual_rate) ** years
    return round(future_value, 2)

print("All defaults:", calculate_future_value())
print("All positional:", calculate_future_value(200, 0.04, 5))
print("All keywords:", calculate_future_value(years=5, starting_amount=200, annual_rate=0.04))
print("Override one default:", calculate_future_value(years=2))
````

 <strong style="color:#512888;font-size:18px;">11 · Common mistakes and good habits</strong><br><span style="color:#38294e;font-size:13px;">Avoid frequent errors in first-year Python programs.</span></div>

| Common mistake | Better approach |
|:---|:---|
| Forgetting the `:` after `def` | End the function header with `:` |
| Not indenting the function body | Indent the body consistently (usually four spaces) |
| Writing a function but never calling it | Use `function_name(...)` to execute it |
| Confusing `print` and `return` | Return reusable results; print only for display |
| Passing arguments in the wrong order | Check the parameter list or use keyword arguments |
| Calling `math.sqrt` without importing `math` | Run `import math` first |
| Assuming every function must take parameters | Functions may have zero parameters |
| Assuming a function without `return` returns its printed text | Such a function returns `None` |

 <strong style="color:#512888;font-size:18px;">12 · Lecture recap</strong><br><span style="color:#38294e;font-size:13px;">Keep these ideas in mind when you write your own functions.</span></div>

**Remember:**

- Functions are reusable blocks of code that execute when called.
- Built-in functions are available immediately; module functions typically require an import.
- `def` creates a function; **parameters** appear in the definition and **arguments** appear in calls.
- `return` makes a computed value available to other code, whereas `print()` displays information.
- Functions can have zero or more parameters, can return useful values, and can call other functions.
- Keyword arguments improve clarity; default values make some arguments optional.
