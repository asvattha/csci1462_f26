<h2 style="margin:0;font-size:28px;color:white;">➡️ Sequential Execution and Decisions</h2>
<p style="margin:8px 0 0;font-size:15px;line-height:1.5;opacity:.95;">Python normally runs statements from top to bottom, but conditions can change that path.</p>
</div>

When Python executes a program, statements normally run **one line after another in sequential order**.

For example:

```python
x = 5
y = 10
total = x + y
print(total)
```

Python starts with the first statement and moves downward one statement at a time.

Sometimes, however, we want Python to **skip certain statements** or choose a different action depending on a condition.

### A real-world analogy

Before driving to the university, you might check the weather:

- If it is **sunny**, you take your sunglasses.
- If it is **raining**, you take your umbrella.

Your course of action changes depending on the weather condition. Python can make similar decisions using
**conditional statements**.

<div style="background:linear-gradient(135deg,#4f1d7f 0%,#7c3aed 100%);color:white;padding:22px 26px;border-radius:16px;margin:34px 0 18px;box-shadow:0 8px 22px rgba(79,29,127,.18);">

<h2 style="margin:0;font-size:28px;color:white;">✅ if Statements</h2>
<p style="margin:8px 0 0;font-size:15px;line-height:1.5;opacity:.95;">Execute a block of code only when a Boolean condition is true.</p>
</div>

The general syntax is:

```python
if boolean_expression:
    statements
```

The indented statements execute only when the Boolean expression evaluates to `True`.

````python
temperature = 85

if temperature > 80:
    print("It is hot outside.")
````

If the condition is `False`, Python skips the indented block.

````python
temperature = 70

if temperature > 80:
    print("It is hot outside.")

print("Have a nice day!")
````

<div style="background:#fffbeb;border:1px solid #fde68a;border-left:5px solid #d97706;padding:17px 20px;border-radius:12px;margin:16px 0 20px;">
<strong>⚠️ Indentation matters</strong><br><div style="margin-top:7px;line-height:1.6;">Python uses indentation to identify the statements that belong to an <code>if</code> block.</div>
</div>

<div style="background:linear-gradient(135deg,#4f1d7f 0%,#7c3aed 100%);color:white;padding:22px 26px;border-radius:16px;margin:34px 0 18px;box-shadow:0 8px 22px rgba(79,29,127,.18);">

<p style="margin:8px 0 0;font-size:15px;line-height:1.5;opacity:.95;">Choose between two possible paths.</p>

Use `if-else` when one action should occur if a condition is `True` and another action should occur if it is `False`.

```python
if boolean_expression:
    statement1
else:
    statement2
```

````python
is_raining = True

if is_raining:
    print("Take an umbrella.")
else:
    print("You do not need an umbrella.")
````

````python
score = 68

if score >= 70:
    print("You passed.")
else:
    print("You did not pass.")
````

<h2 style="margin:0;font-size:28px;color:white;">🛣️ if-elif-else Statements</h2>
<p style="margin:8px 0 0;font-size:15px;line-height:1.5;opacity:.95;">Choose one path from several possible alternatives.</p>
</div>

When a problem has more than two possible outcomes, use an `if-elif-else` chain.

```python
if condition1:
    statement1
elif condition2:
    statement2
elif condition3:
    statement3
else:
    statement4
```

Python checks conditions **from top to bottom**. As soon as it finds the **first condition that is `True`**,
it executes that block and skips the rest of the chain.

````python
temperature = 75

if temperature >= 90:
    print("Very hot")
elif temperature >= 80:
    print("Hot")
elif temperature >= 70:
    print("Comfortable")
else:
    print("Cool")
````

<div style="background:#f0fdf4;border:1px solid #bbf7d0;border-left:5px solid #16a34a;padding:17px 20px;border-radius:12px;margin:16px 0 20px;">
<strong>⭐ Key rule</strong><br><div style="margin-top:7px;line-height:1.6;">An <code>if-elif-else</code> chain executes the block associated with the <strong>first true condition</strong>.</div>
</div>

## Why Separate `if` Statements Can Cause Problems

Suppose we want to print one letter grade:

````python
score = 95

if score >= 90:
    print("A")

if score >= 80:
    print("B")

if score >= 70:
    print("C")

if score >= 60:
    print("D")

if score < 60:
    print("F")
````

For `score = 95`, the first four conditions are all true, so the code prints:

```text
A
B
C
D
```

Each separate `if` statement is evaluated independently.

## Better Solution Using `if-elif-else`

````python
score = 95

if score >= 90:
    print("A")
elif score >= 80:
    print("B")
elif score >= 70:
    print("C")
elif score >= 60:
    print("D")
else:
    print("F")
````

Now Python prints only:

```text
A
```

Once `score >= 90` is true, Python skips the remaining branches.

## Why the Order of Conditions Matters

The following order is incorrect for letter grades:

````python
score = 95

if score >= 60:
    print("D")
elif score >= 70:
    print("C")
elif score >= 80:
    print("B")
elif score >= 90:
    print("A")
else:
    print("F")
````

This prints `D`, because `95 >= 60` is already true.

For threshold-based grading, test the **highest threshold first**, then move downward.

<h2 style="margin:0;font-size:28px;color:white;">🔁 Loops: Repeating Actions</h2>
<p style="margin:8px 0 0;font-size:15px;line-height:1.5;opacity:.95;">Loops repeat a block of code instead of forcing us to copy the same statements many times.</p>
</div>

Another Python construct that changes normal line-by-line execution is a **loop**.

In everyday life, we often repeat actions. For example, from Monday through Thursday you might repeatedly:

1. Wake up in the morning.
2. Brush your teeth and get ready.
3. Commute to the university.

Programs also need repetition. Instead of copying and pasting the same code many times, we use loops.

Suppose we want to print `"Hello"` ten times.

Instead of writing `print("Hello")` ten times, we can write:

````python
for i in range(10):
    print("Hello")
````

<div style="background:#eef6ff;border:1px solid #cfe4ff;border-left:5px solid #2563eb;padding:17px 20px;border-radius:12px;margin:16px 0 20px;">
<strong>💡 What is a loop?</strong><br><div style="margin-top:7px;line-height:1.6;">A <strong>loop</strong> repeatedly executes a block of code. A <code>for</code> loop is especially useful when we want to process every item in a collection or repeat an action a known number of times.</div>
</div>

<div style="background:linear-gradient(135deg,#4f1d7f 0%,#7c3aed 100%);color:white;padding:22px 26px;border-radius:16px;margin:34px 0 18px;box-shadow:0 8px 22px rgba(79,29,127,.18);">
<div style="font-size:13px;font-weight:700;letter-spacing:.7px;opacity:.9;margin-bottom:5px;">SECTION 6</div>
<h2 style="margin:0;font-size:28px;color:white;">🔄 The for Loop</h2>
<p style="margin:8px 0 0;font-size:15px;line-height:1.5;opacity:.95;">Visit the items of an iterable one at a time.</p>
</div>

The general syntax is:

```python
for variable_name in iterable:
    statements
```

For every item in the iterable:

1. Python stores the current item in `variable_name`.
2. Python executes the indented loop body.
3. Python gets the next item.
4. The process repeats until no items remain.

<div style="background:#f7f4ff;border:1px solid #ddd0f7;border-left:5px solid #7c3aed;padding:17px 20px;border-radius:12px;margin:16px 0 20px;">
<strong>📌 What is an iterable?</strong><br><div style="margin-top:7px;line-height:1.6;">An <strong>iterable</strong> is an object whose elements can be visited one at a time. Common iterables include strings, lists, tuples, dictionaries, and <code>range</code> objects.</div>
</div>

### Looping through a list

````python
numbers = [10, 20, 30]

for number in numbers:
    print(number)
````

### Looping through a string

````python
word = "DATA"

for letter in word:
    print(letter)
````

### Looping through a tuple

````python
coordinates = (2, 5, 8)

for value in coordinates:
    print(value)
````

### Looping through a dictionary

By default, looping over a dictionary visits its **keys**.

````python
student = {
    "name": "Alice",
    "score": 92
}

for key in student:
    print(key)
````

<h2 style="margin:0;font-size:28px;color:white;">🔢 The range() Function</h2>
<p style="margin:8px 0 0;font-size:15px;line-height:1.5;opacity:.95;">Generate a sequence of integers for counting and repetition.</p>

`range()` creates an iterable sequence of integers and is commonly used with `for` loops.

```python
range(stop)
range(start, stop)
range(start, stop, step)
```

The `stop` value is **not included**.

### `range(stop)`

````python
for i in range(5):
    print(i)
````

This produces:

```text
0
1
2
3
4
```

### `range(start, stop)`

````python
for i in range(2, 6):
    print(i)
````

### `range(start, stop, step)`

````python
for i in range(0, 10, 2):
    print(i)
````

### Counting backward with a negative step

````python
for i in range(5, 0, -1):
    print(i)
````

<h2 style="margin:0;font-size:28px;color:white;">🧮 Tracing a for Loop</h2>
<p style="margin:8px 0 0;font-size:15px;line-height:1.5;opacity:.95;">Follow every important variable from one iteration to the next.</p>

Consider a loop that adds all elements in a list:

````python
total = 0

for item in [1, 2, 3, 4, 5]:
    total += item

print(total)
````

It is a good idea to **trace** a loop to understand how Python executes it.

| Loop Iteration | `item` | Calculation | New `total` |
|---:|---:|---|---:|
| Before loop | — | — | `0` |
| 1 | `1` | `0 + 1` | `1` |
| 2 | `2` | `1 + 2` | `3` |
| 3 | `3` | `3 + 3` | `6` |
| 4 | `4` | `6 + 4` | `10` |
| 5 | `5` | `10 + 5` | `15` |

At the end of the loop, `total` contains the sum of every element: `15`.

### A Loop Can Contain a Conditional Statement

````python
even_count = 0

for number in [4, 7, 10, 13, 18]:
    if number % 2 == 0:
        even_count += 1

print(even_count)
````

The loop visits every number, but the counter changes only when the `if` condition is true.

<h2 style="margin:0;font-size:28px;color:white;">🧱 Nested Loops</h2>
<p style="margin:8px 0 0;font-size:15px;line-height:1.5;opacity:.95;">Place one loop inside another when repetition itself must be repeated.</p>
</div>

A **nested loop** is a loop inside another loop.

```python
for outer_variable in outer_iterable:
    for inner_variable in inner_iterable:
        statements
```

For each **one iteration of the outer loop**, the inner loop runs through all of its iterations.

Suppose we want to print:

```text
*****
*****
*****
```

There are 3 rows and 5 stars in each row.

- The **outer loop** controls the rows.
- The **inner loop** controls the stars within a row.

````python
for row in range(3):
    s = ""
    for column in range(5):
        s += "*"
    print(s)
````

<strong>🧠 Mental model for nested loops</strong><br><div style="margin-top:7px;line-height:1.6;">Think of the <strong>outer loop as rows</strong> and the <strong>inner loop as columns</strong>. This mental model is especially useful when working with matrices and tabular data.</div>
</div>

## Tracing the Nested Loop

For:

```python
for row in range(3):
    s = ""
    for column in range(5):
        s += "*"
    print(s)
```

### Outer iteration 1

`row = 0`

The inner loop takes:

```text
column = 0
column = 1
column = 2
column = 3
column = 4
```

Five stars are stored in the varaible `s`:

```text
*****
```

Then `print(s)` prints the five stars and move to next line.

### Outer iteration 2

`row = 1`

The inner loop again runs five times and prints:

```text
*****
```

### Outer iteration 3

`row = 2`

The inner loop runs five more times and prints:

```text
*****
```

Therefore, the final output is:

```text
*****
*****
*****
```

### How many times does the inner statement execute?

The outer loop runs `3` times and the inner loop runs `5` times for each outer iteration.

```text
3 × 5 = 15
```

Therefore:

```python
s += "*"
```

executes **15 times**.

````python
count = 0

for row in range(3):
    for column in range(5):
        count += 1

print("Inner-loop executions:", count)
````

<h2 style="margin-top:0;">✅ Key Takeaways</h2>

### Conditional Statements

- Python normally executes statements from **top to bottom**.
- `if` runs code only when a condition is `True`.
- `if-else` chooses between two alternatives.
- `if-elif-else` chooses one path from several alternatives.
- An `if-elif-else` chain stops after the **first true condition**.
- The **order of conditions matters**.

### Loops

- A loop repeats a block of code.
- A `for` loop visits items from an **iterable** one at a time.
- Strings, lists, tuples, dictionaries, and `range` objects are common iterables.
- `range()` is useful when repeating code a known number of times.
- The stop value of `range()` is **excluded**.
- An **accumulator** stores a running result, such as a sum.
- A **counter** tracks how many times something occurs.
- Tracing a loop iteration-by-iteration is an important debugging skill.
- A **nested loop** is a loop inside another loop.
- If the outer loop runs `m` times and the inner loop runs `n` times each time, the inner body executes `m × n` times.
