<strong>📐 Formula 1 — Energy Loss</strong>
<div style="margin-top:10px;font-size:18px;text-align:center;background:white;padding:12px;border-radius:8px;">
Energy Loss = Energy Supplied − Energy Billed
</div>
</div>

<div style="background:#f5f3ff;padding:18px;border-radius:10px;border:1px solid #ddd6fe;margin:15px 0 25px 0;">
<strong>📐 Formula 2 — Loss Percentage</strong>
<div style="margin-top:10px;font-size:18px;text-align:center;background:white;padding:12px;border-radius:8px;">
Loss Percentage = (Energy Loss ÷ Energy Supplied) × 100
</div>
</div>
