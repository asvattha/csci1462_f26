# Numpy Arrays


## What is NumPy?

NumPy is a Python library that is mainly used for numerical and scientific computing.

Its main data structure is the **NumPy array**.

NumPy arrays are usually created using the `array()` function:

```python
numpy.array([...])
```

NumPy arrays are similar to Python lists because they allow us to store multiple values in an ordered sequence.

However, NumPy arrays have several important differences from Python lists.

We will explore these differences throughout this tutorial.

First make sure you import the `numpy` module to use numpy arrays. 


````python
import numpy as np
````

## Creating a NumPy Array

We can create a NumPy array using the `np.array()` function.

For example:


````python
numbers = np.array([10, 20, 30, 40, 50])

numbers
````

However, this is a NumPy array, not a Python list.

We can check its type using type().

````python
type(numbers)
````

**ndarray** stands for N-dimensional array.

This is the main data structure provided by NumPy.


## NumPy Arrays vs Python Lists

NumPy arrays and Python lists may look similar, but they behave differently.

Let's first create a Python list and a NumPy array containing the same values.


````python
my_list = [1, 2, 3, 4, 5]

my_array = np.array([1, 2, 3, 4, 5])

my_list, my_array
````

Both contain the same values, but they are different data structures.

Let's check their types.

````python
type(my_list), type(my_array)
````

The list is a Python `list`, while the NumPy object is a `numpy.ndarray`.

One of the biggest advantages of NumPy arrays is that they are designed specifically for numerical operations. Consider the following code which multiplies each number between 0 to 999999 using numpy methods and traditional for loop and python lists. 

````python
# you do not need to know how this code works. Keep reading the next cell
import time 

numbers_list = list(range(1_000_000))

start = time.time()
result = []
for number in numbers_list:
    result.append(number * 2)
end = time.time()
list_time = end - start


numbers_array = np.arange(1_000_000)
start = time.time()
result = numbers_array * 2
end = time.time()
numpy_time = end - start

print("List time:", list_time, "secs")
print("NumPy time:", numpy_time, "secs")
````

Your exact results will depend on your computer and what else is running on it.

However, NumPy will generally be much faster for large numerical operations because its operations are implemented using highly optimized compiled code.

This is especially important in data science because datasets can contain millions or even billions of values.

Instead of repeatedly writing Python `for` loops, we can often use NumPy's vectorized operations to process the data much more efficiently.

# Features of Numpy Array

## 1. Homogeneous Data Types

NumPy arrays have a single data type, known as their **dtype**.

For example, if we create an array containing integers:

````python
numbers = np.array([10, 20, 30, 40])

numbers
````

````python
numbers.dtype
````

The `dtype` tells us the type of data stored in the array. In the above example, numbers stores `int` type of values, and each value takes 64 bits in the memory.

We can also create an array containing real numbers.

````python
prices = np.array([10.5, 20.75, 30.25])

prices
````

````python
prices.dtype
````

Python lists can contain different types of data:

```python
my_list = [10, "hello", 3.5, True]
```

NumPy arrays, on the other hand, use a common data type for their elements.

````python
mixed = np.array([10, "hello", 3.5, True])

mixed
````

Notice that NumPy converted the integers, floats, booleans to string so that all elements could have the same data type.

This consistency allows NumPy to store and process data very efficiently.

## 2. NumPy Arrays Are Mutable

NumPy arrays are **mutable**, which means that we can change the values of existing elements.

For example:

````python
numbers = np.array([10, 20, 30, 40])

numbers
````

````python
numbers[0] = 100

numbers
````

The first element changed from `10` to `100`.

So NumPy arrays are mutable.

However, NumPy arrays generally have a **fixed size and shape** after they are created. This is different from Python lists, where we can easily append new elements.

## 3. Indexing and Slicing NumPy Arrays

Like Python lists, NumPy arrays use indexing.

Remember that Python indexing starts at `0`.

For example:

````python
numbers = np.array([10, 20, 30, 40, 50])

numbers[0]
````

````python
numbers[2]
````

````python
numbers[-1]
````

We can also use slicing, just like we do with Python lists.

````python
numbers[1:4]
````

````python
numbers[:3]
````

````python
numbers[2:]
````

## 4. Shape and Dimensions

One of the most important features of NumPy arrays is that they can have multiple dimensions.

A simple array such as:

```python
[10, 20, 30, 40]
```

is a one-dimensional array.

````python
numbers = np.array([10, 20, 30, 40])
numbers.ndim
````

The `ndim` attribute tells us the number of dimensions.

Our array has:

```text
1 dimension
```

We can also check its shape.

````python
numbers.shape
````

This tells us that the array contains 4 elements along its only dimension.

Now let's create a two-dimensional array.

````python
matrix = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

matrix
````

````python
matrix.ndim
````

````python
matrix.shape
````

NumPy uses the shape of an array to understand how its data is organized. Here, (2, 3) means it's a 2-dimensional array, where there are two rows, and each row has 3 elements/columns. Each element in a NumPy array has a well-defined shape, which is crucial for performing vectorized operations. For instance, in a 2D array (matrix), all rows have the same number of columns, ensuring consistent dimensionality.

## 5. Accessing Elements in a 2D Array

For a two-dimensional array, we provide two indexes:

```python
array[row, column]
```

````python
matrix[0, 0]
````

````python
matrix[1, 1] # select the item at 2nd row and 2nd col of the matrix
````

````python
matrix[1, 1:] # select the second row, and columns starting from 2nd column to the very end
````

````python
matrix[1] # select the second row of the matrix
````

````python
matrix[:, 2] # select the third column of the matrix
````

````python
matrix[[0, 1], [2, 2]] # select the items at index (0, 2) and (1, 2) from the matrix 
````

# Why Do Data Scientists Use NumPy?

Many operations that we perform with NumPy can also be performed using Python lists and `for` loops.

So why do we use NumPy?

There are several reasons:

### 1. Speed

NumPy is highly optimized for numerical operations and can process large amounts of data very quickly.

### 2. Vectorized Operations

We can perform operations on entire arrays without explicitly writing `for` loops.

### 3. Memory Efficiency

NumPy arrays can store numerical data more efficiently than regular Python lists.

### 4. Multidimensional Data

NumPy makes it easy to work with one-dimensional, two-dimensional, and higher-dimensional data.

### 5. Foundation for Data Science

Many Python libraries used in data science and machine learning rely heavily on NumPy.

For example, pandas uses NumPy internally for many of its operations.

This is why learning NumPy is an important step toward learning data analysis.
