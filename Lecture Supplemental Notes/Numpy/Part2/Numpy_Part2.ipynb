### Scalar computation on a numpy

A **scalar** is one value, such as `2`, `-5`, or `0.25`. When a scalar and a NumPy array are used in an arithmetic expression, NumPy applies the operation to **every element**.

This behavior is a simple form of **broadcasting** (we will learn in our next lesson). NumPy conceptually stretches the scalar to match the array.

| Expression | Calculation performed |
|---|---|
| `values + 3` | Add 3 to every element |
| `values - 3` | Subtract 3 from every element |
| `values * 3` | Multiply every element by 3 |
| `values / 3` | Divide every element by 3 |
| `values ** 2` | Square every element |

<img src="pics/broadcast1.jpeg" width=500px />

````python
arr = np.array([1, 2, 3])
arr * 2
````

````python
values = np.array([1, 2, 3, 4])

print("Original:  ", values)
print("Times 2:   ", values * 2)
print("Plus 10:   ", values + 10)
print("Squared:   ", values ** 2)
print("Divided by 2:", values / 2)
````

<div class="np-card np-tip">
<h4>Why this matters</h4>

The expression `values * 2` replaces a loop such as “visit each position and multiply its value by 2.” This style is called a **vectorized operation**. It is concise and is usually faster than a Python loop for large arrays.
</div>

The original array is unchanged unless we assign a result back to it or another variable.

### Computation between two numpy arrays

When two arrays have the same shape, arithmetic is performed **element by element**. The value at index `0` is combined with the value at index `0`, the value at index `1` with the value at index `1`, and so on.

````python
a = np.array([10, 20, 30])
b = np.array([1, 2, 3])

print("a + b =", a + b)
print("a - b =", a - b)
print("a * b =", a * b)
print("a / b =", a / b)
````

## Creating sequences

### `np.arange()`

`np.arange()` creates evenly spaced values using a familiar start–stop–step pattern:

```python
np.arange(start, stop, step)
```

- `start` is included.
- `stop` is excluded.
- `step` controls the distance between consecutive values.
- With one argument, NumPy uses `start=0` and `step=1`.


````python
print("One argument:   ", np.arange(5))
print("Start and stop:", np.arange(3, 9))
print("With a step:   ", np.arange(3, 30, 5))
print("Counting down: ", np.arange(5, 0, -1))
````

The stop value is excluded. For example, `np.arange(3, 9)` ends at `8`, and `np.arange(5, 0, -1)` ends at `1`.

````python
np.arange(10) # here start = 0 and step = 1 are set by default
````

````python
np.arange(3, 9) # here start = 3, end = 8, step = 1 by default
````

````python
np.arange(3, 30, 5) # start = 3, end = 29, step = 5 
````

````python
np.arange(1.5, -2, -0.5) # start = 1.5, end = -1.99999.... , step = -0.5
````

### Summarizing an array

Functions that combine many values into one summary are often called **reductions**.

<div class="np-grid">
  <div class="np-mini"><strong><code>np.sum(a)</code></strong><br>Total</div>
  <div class="np-mini"><strong><code>np.mean(a)</code></strong><br>Arithmetic mean</div>
  <div class="np-mini"><strong><code>np.min(a)</code></strong><br>Smallest value</div>
  <div class="np-mini"><strong><code>np.max(a)</code></strong><br>Largest value</div>
  <div class="np-mini"><strong><code>np.prod(a)</code></strong><br>Product of all values</div>
</div>


````python
temperatures = np.array([71, 74, 79, 76, 72])

print("Total:  ", np.sum(temperatures))
print("Mean:   ", np.mean(temperatures))
print("Minimum:", np.min(temperatures))
print("Maximum:", np.max(temperatures))
print("Product of [2, 3, 4]:", np.prod(np.array([2, 3, 4])))
````

````python
arr = np.array([1, 2, 3])
arr = np.arange(5)
np.sum(arr) # add all elements of an array
````

````python
np.mean(arr) # returns the mean of all the elements of the array
````

````python
np.prod(arr) # returns the product of all the elements of the array
````

## Boolean array

When all the elements of a numpy array contains `True` or `False` values, then it's a boolean array.

Three functions are especially useful with Boolean arrays:

- `np.all(condition)` is `True` only if **every** element is `True`.
- `np.any(condition)` is `True` if **at least one** element is `True`.
- `np.count_nonzero(condition)` counts the `True` values.


````python
scores = np.array([88, 67, 91, 73, 59])
passed = scores >= 70

print("Scores: ", scores)
print("Passed? ", passed)
````

````python
print("Did every student pass?", np.all(passed))
print("Did any student pass?  ", np.any(passed))
print("Number who passed:     ", np.count_nonzero(passed))
````

These functions also accept numerical arrays. NumPy treats zero as `False` and a nonzero value as `True`.

````python
arr = np.array([0, 5, 3, 0])
np.all(arr) # returns True if all elements are True or non-zero values
````

````python
barr = np.array([True, True, True])
np.all(barr)
````

````python
barr = np.array([True, True, True, False])
np.all(barr)
````

````python
arr = np.array([0, 5, 3, 0])
np.any(arr) # returns True if any of the elements are True or non-zero value 
````

````python
barr = np.array([False, False, False])
np.any(barr)
````

````python
barr = np.array([False, False, False, True])
np.any(barr)
````

````python
arr = np.array([0, 5, 3, 0, 8])
np.count_nonzero(arr) # counts the number of True or non-zero elements in the array
````

## Transforming arrays

The following functions return new arrays rather than reducing the data to one number.

### Consecutive differences: `np.diff()`

`np.diff()` subtracts each element from the element immediately after it:

$$[x_0,x_1,x_2,x_3] \longrightarrow [x_1-x_0,\;x_2-x_1,\;x_3-x_2]$$

Because it measures gaps between adjacent values, the result has one fewer element.


````python
np.diff(arr) # returns the difference of the consecutive elements 
````

### Rounding: `np.round()`

Use the second argument to choose the number of digits after the decimal point.


````python
arr = np.array([0.5, 0.239, 0.8238])
np.round(arr, 2) # returns the round of each number to the nearest integer
````

### Running totals and products

- `np.cumsum()` produces a running total.
- `np.cumprod()` produces a running product.

````python
arr = np.array([1, 2, 3, 4])
np.cumprod(arr) # a cumulative product
````

````python
np.cumsum(arr) # a cumulative sum
````

### Common mathematical transformations

| Function | Meaning | Input caution |
|---|---|---|
| `np.exp(a)` | $e^x$ for each value $x$ | Can grow very quickly |
| `np.log(a)` | Natural logarithm, $\ln(x)$ | Use positive inputs |
| `np.sqrt(a)` | Square root | Use nonnegative inputs for real results |

````python
np.exp(arr) # exponentiate each element 
````

````python
np.log(arr) # natural logarithm of each element
````

````python
np.sqrt(arr) # square root of each element
````

### Sorting: `np.sort()`

`np.sort(a)` returns a sorted copy; it does not change `a`.


````python
np.sort(arr) # sort elements in ascending order
````

### Finding the position of an extreme value

`np.argmax()` returns the index of the first maximum value, while `np.argmin()` returns the index of the first minimum value.


````python
temperatures = np.array([71, 74, 79, 76, 72])

hottest_index = np.argmax(temperatures)
coldest_index = np.argmin(temperatures)

print("Index of first maximum:", hottest_index)
print("Maximum temperature:    ", temperatures[hottest_index])
print("Index of first minimum:", coldest_index)
print("Minimum temperature:    ", temperatures[coldest_index])
````
