# 1. Comments in Python

Comments are notes written for **people**, not instructions for Python. The interpreter ignores comments when the program runs.

Use comments to explain **why** something is being done, clarify a non-obvious step, or temporarily disable a line while testing.

### Single-line comments

A single-line comment begins with `#`.

````python
# This is a full-line comment.
temperature = 92

# An inline comment appears after executable code.
is_hot = temperature > 90  # Check whether the temperature exceeds 90°F

print(is_hot)
````

### Good comments vs. unnecessary comments

A useful comment adds information that the code does not already make obvious.

```python
# Convert the temperature from Fahrenheit to Celsius for comparison with the dataset.
celsius = (fahrenheit - 32) * 5 / 9
```

The following comment adds very little:

```python
# Add 1 to count
count = count + 1
```

Aim to explain the **reason or intent**, not simply repeat the code in English.

## Docstrings are not the same as comments

Python does **not** have a special multiline-comment syntax.

Triple-quoted strings such as `""" ... """` create **string literals**. When placed immediately inside a function, class, or module, they become **docstrings** and can be accessed by Python.

Docstrings are commonly used to document what a function does, its parameters, and its return value.

````python
def double(x):
    """
    x: a numerical value
    Return twice the value of x.
    """
    return 2 * x

print(double(6))
print(double.__doc__)
````

### Temporarily commenting out code

During debugging, you can place `#` at the beginning of a line to prevent that line from running.

````python
# print("Hello")
# print("World")

print("Hello World")
````

<h2 style="margin-top:0;">⚠️ 2. Errors in Python</h2>

Python problems usually fall into three broad categories:

| Category | What happens? | Typical example |
|---|---|---|
| **Syntax error** | Python cannot understand the code's structure. | Missing quote, parenthesis, colon, or incorrect indentation |
| **Runtime error / exception** | The code is valid Python, but something goes wrong while it runs. | Undefined variable, invalid type, missing file |
| **Logical error** | The program runs, but produces the wrong result. | Using the wrong formula |

</div>

## How to read a traceback

A traceback often looks intimidating, but you usually need only a few clues:

1. **Look at the last line** — it tells you the error type and usually gives a short explanation.
2. **Find the line of your code** referenced just above it.
3. **Read the message carefully** and inspect the variables used on that line.
4. Fix one issue at a time and run the cell again.

For example:

```text
NameError: name 'totl' is not defined
```

The important parts are:

- `NameError` → the category of the problem.
- `totl` → the name Python could not find.
- `is not defined` → likely a typo or a variable that was never assigned.

## 2.1 `SyntaxError`

A `SyntaxError` occurs when Python cannot parse the code because it violates Python's grammar.

Common causes include:

- Missing quotation marks
- Missing parentheses
- Missing `:` after `if`, `for`, `while`, or `def`
- Using `=` where a comparison such as `==` is required inside a condition
- Other malformed Python statements

The example below *demonstrates* a syntax error safely using `compile()` so the notebook does not stop.

````python
name = "Dipak Singh'
````

### `IndentationError`

`IndentationError` is closely related to `SyntaxError`. Python uses indentation to define blocks of code, so inconsistent or missing indentation can prevent the program from running.

````python
x = 5
  y = 25
````

## 2.2 `NameError`

A `NameError` occurs when Python encounters a name that has not been defined.

This is extremely common when:

- A variable name is misspelled.
- A cell that creates the variable was never run.
- The kernel was restarted and variables were cleared.

In Jupyter notebooks, **cell execution order matters**.

````python
score = 90
print(socre)
````

## 2.3 `TypeError`

A `TypeError` occurs when an operation is applied to an object of an inappropriate type.

A common beginner example is mixing a string and a number in arithmetic.

````python
age = "20"

next_age = age + 1

# One possible fix:
next_age = int(age) + 1
print("Corrected next age:", next_age)
````

## 2.4 `ValueError`

A `ValueError` occurs when the **type is acceptable**, but the actual value cannot be used in the requested operation.

This is common when cleaning data or converting text to numbers.

````python
value = "twenty"

number = int(value)

# A valid conversion:
number = int("20")
print("Successful conversion:", number)
````

## 2.5 `ZeroDivisionError`

A `ZeroDivisionError` occurs when you attempt to divide by zero.

In data analytics, this can happen when calculating an average, rate, percentage, or ratio and the denominator happens to be zero.

````python
total_sales = 500
number_of_customers = 0

total_sales / number_of_customers
````

## 2.6 `IndexError`

An `IndexError` occurs when you try to access a sequence position that does not exist.

Remember that Python indexing begins at `0`.

````python
scores = [80, 90, 100]

print("Valid last item:", scores[5])
````

## 2.7 `KeyError`

A `KeyError` occurs when you ask a dictionary—or commonly a pandas `DataFrame`—for a key or column label that does not exist.

This is especially common in data analytics because column names must match exactly, including spaces and capitalization.

````python
student = {"name": "Alex", "score": 92}

student["age"]
````

### Data analytics example: pandas column names

If a DataFrame contains a column named `"Sales"` but you write:

```python
df["sales"]
```

pandas will usually raise a `KeyError` because `"Sales"` and `"sales"` are different strings.

A useful debugging step is:

```python
print(df.columns)
```

to inspect the exact column names.

## 2.8 `AttributeError`

An `AttributeError` occurs when you try to use a method or attribute that an object does not have.

This often happens when you remember the right operation but apply it to the wrong type of object.

````python
numbers = [10, 20, 30]
numbers.upper()
````

## 2.9 `FileNotFoundError`

A `FileNotFoundError` occurs when Python cannot find a requested file.

This is one of the most common errors when loading datasets.

Typical causes:

- The filename is misspelled.
- The file is in a different folder.
- The notebook's current working directory is not what you expected.
- The file extension is incorrect (`.csv` vs `.xlsx`, for example).

````python
from pathlib import Path
import pandas as pd

filename = "sample_data_that_does_not_exist.csv"

pd.read_csv(filename)
````

## 2.10 `ModuleNotFoundError` / `ImportError`

These errors are related to importing libraries.

- `ModuleNotFoundError` usually means Python cannot find the package you asked to import.
- `ImportError` can occur when a package exists but the requested object cannot be imported from it.

In a data analytics course, you may see these when working with libraries such as `pandas`, `numpy`, or `matplotlib`.

If a package is truly missing, install it in the appropriate environment before importing it. Do **not** repeatedly install packages every time a notebook runs unless your course environment specifically requires that.

````python
import mth
````

## 2.11 Logical Errors

A **logical error** is different from all the above errors: the program runs successfully, but the result is incorrect.

Python cannot always detect these because the code is syntactically valid and the operations themselves are legal.

Logical errors often require you to:

- Check the formula.
- Test with values whose answer you already know.
- Print intermediate values.
- Compare the code with the problem statement.

````python
side_of_square = 5

# Logical error: this calculates the volume-like value 5^3, not square area.
incorrect_area = side_of_square ** 3
print("Incorrect result:", incorrect_area)

# Correct formula:
correct_area = side_of_square ** 2
print("Correct area:", correct_area)
````

<h3 style="margin-top:0;">🧭 Common Error Quick Reference</h3>

| Error | Meaning | First thing to check |
|---|---|---|
| `SyntaxError` | Python cannot understand the code structure. | Quotes, parentheses, colons, operators |
| `IndentationError` | A code block is indented incorrectly. | Spaces/tabs and block alignment |
| `NameError` | A name has not been defined. | Spelling and whether the defining cell ran |
| `TypeError` | An operation used incompatible types. | `type(variable)` and conversions |
| `ValueError` | The type is okay, but the value is invalid. | The actual contents of the variable |
| `ZeroDivisionError` | Division by zero was attempted. | Denominator |
| `IndexError` | A sequence index is out of range. | `len(sequence)` and zero-based indexing |
| `KeyError` | A key/column name does not exist. | Dictionary keys or `df.columns` |
| `AttributeError` | The object has no such method/attribute. | Object type and method spelling |
| `FileNotFoundError` | Python cannot locate a file. | Filename, path, working directory |
| `ModuleNotFoundError` | Python cannot locate an imported package. | Package installation/environment |
| Logical error | Code runs but produces the wrong answer. | Formula and reasoning |

</div>

# 3. A Simple Debugging Strategy

When something goes wrong, use this process:

### Step 1 — Read the last line of the traceback
Identify the error type: `NameError`, `TypeError`, `KeyError`, etc.

### Step 2 — Locate the failing line
Look for the line number or notebook cell shown in the traceback.

### Step 3 — Inspect the variables
Useful tools include:

```python
print(variable)
type(variable)
len(variable)
```

### Step 4 — Fix one problem at a time
Do not change many unrelated lines at once.

### Step 5 — Run the cell again
In Jupyter, also make sure prerequisite cells have been executed in the correct order.
