## 2.1 What is a Pandas Series?

A **Pandas Series** is a one-dimensional collection of values.

Each value can be accessed in two different ways:

1. By its **index label** (named index)
2. By its **integer position** (numbered index)

If you do not supply index labels, Pandas creates labels `0, 1, 2, ...`.

````python
scores = pd.Series([85, 92, 78, 95])
scores
````

The Series has four values. By default, its labels happen to look exactly like its positions i.e, `0`, `1`, `2` and `3`.

Let's inspect the index and the values separately.

````python
print("Index labels:", scores.index)
print("Values:", scores.values)
````

<strong>⚠️ Label vs. position</strong><br>
<div style="margin-top:7px; line-height:1.6;">
A Series always has integer <strong>positions</strong> <code>0, 1, 2, ...</code>. Its <strong>index labels</strong> can instead be names, dates, numbers, or other values. Do not assume a label and a position are the same thing.
</div>
</div>

## 2.2 Creating a Series with Named Index Labels

Use the `index=` argument to provide your own labels.

````python
student_scores = pd.Series(
    [85, 92, 78, 95],
    index=["Alice", "Bob", "Charlie", "David"]
)

student_scores
````

The Series now has:

| Position | Index label | Value |
|---:|---|---:|
| `0` | `"Alice"` | `85` |
| `1` | `"Bob"` | `92` |
| `2` | `"Charlie"` | `78` |
| `3` | `"David"` | `95` |

## 2.3 Setting or Replacing a Series Index

You can assign a new sequence of labels to `.index`.

````python
temperatures = pd.Series([72, 75, 70, 68])
temperatures
````

````python
temperatures.index = ["Monday", "Tuesday", "Wednesday", "Thursday"]
temperatures
````

## 2.4 Creating a Series from a Dictionary

A Python dictionary maps keys to values.

When a dictionary is converted to a Series:

- dictionary **keys** become index labels
- dictionary **values** become Series values

````python
population_dict = {
    "Texas": 31.3,
    "California": 39.4,
    "Florida": 23.4
}

population = pd.Series(population_dict)
population
````

## 2.5 Duplicate Index Labels

A Pandas Series **may have duplicate index labels**.

````python
regional_sales = pd.Series(
    [120, 150, 90, 175],
    index=["East", "West", "East", "North"]
)

regional_sales
````

## 2.6 Selecting Values with `.loc[]`

`.loc[]` means **locate by label**.

````python
student_scores
````

````python
student_scores.loc["Bob"]
````

You can also request several labels at once by passing a list.

````python
student_scores.loc[["Alice", "David"]]
````

## 2.7 Selecting Values with `.iloc[]`

`.iloc[]` means **integer-location**.

It uses integer positions, beginning with position `0`.

````python
student_scores.iloc[0]
````

````python
student_scores.iloc[2]
````

````python
student_scores.iloc[[0, 3]]
````

## 2.8 Slicing a Series by Numbered Position

Use `.iloc[start:stop]`.

Just like normal Python slicing, the **stop position is excluded**.

````python
student_scores
````

````python
student_scores.iloc[1:3]
````

The expression above includes positions `1` and `2`, but not position `3`.

````python
student_scores.iloc[:2]
````

````python
student_scores.iloc[2:]
````

You can also use a step.

````python
student_scores.iloc[::2]
````

## 2.9 Slicing a Series by Named Index Labels

Use `.loc[start_label:stop_label]`.

With label slicing, Pandas generally **includes the ending label**.

````python
student_scores.loc["Bob":"David"]
````

Python string methods such as `.lower()` and `.strip()` normally operate on **one string**.

For a Pandas Series containing strings, use the **`.str` accessor**:

```python
series.str.method()
```

````python
names = pd.Series([
    "  Alice Smith  ",
    "BOB JONES",
    "charlie brown",
    "Diana Prince"
])

names
````

## 3.1 `str.strip()`

Remove leading and trailing whitespace from every string.

````python
names.str.strip()
````

## 3.2 `str.lower()` and `str.upper()`

````python
names.str.lower()
````

````python
names.str.upper()
````

## 3.3 `str.title()`

Convert each string to title case.

````python
names.str.strip().str.title()
````

## 3.4 `str.replace()`

Replace text inside every string.

````python
names.str.strip().str.replace(" ", "_", regex=False)
````

## 3.5 `str.contains()`

Check each string for a pattern.

The result is another Series containing `True` or `False`.

````python
names.str.contains("alice", case=False)
````

## 3.6 `str.startswith()` and `str.endswith()`

````python
names.str.strip().str.startswith("A")
````

````python
names.str.strip().str.endswith("s")
````

## 3.7 `str.len()`

Count the number of characters in each string.

````python
names.str.strip().str.len()
````

## 3.8 Chaining String Methods

Pandas operations can be chained so that the output of one step becomes the input to the next.

````python
clean_names = (
    names
    .str.strip()
    .str.lower()
    .str.replace(" ", "_", regex=False)
)

clean_names
````

## 4.1 What is a DataFrame?

A **Pandas DataFrame** is a two-dimensional data structure organized into:

- rows
- columns
- row index labels
- column labels

It resembles a spreadsheet or a database table.

## 4.2 Creating a DataFrame from a Dictionary

One of the most common ways to create a DataFrame is from a dictionary.

Each dictionary key becomes a **column name**.

````python
students = pd.DataFrame({
    "Name": ["Alice", "Bob", "Charlie", "David", "Emma"],
    "Major": ["CS", "Math", "CS", "Physics", "Math"],
    "Score": [91, 84, 95, 77, 88],
    "Year": [2, 3, 1, 4, 2]
})

students
````

## 6.6 `.iloc[]` with a DataFrame

The general form is:

```python

```

`.iloc[]` works with **integer positions**.

````python
students.iloc[0]
````

````python
students.iloc[0, 2]
````

`students.iloc[0, 2]` means:

- row position `0`
- column position `2`

Column position `2` is `"Score"`.

## 7.1 Series: Position-Based Slicing

For `.iloc[]`, the stop position is excluded.

````python
example_series = pd.Series(
    [10, 20, 30, 40, 50],
    index=["A", "B", "C", "D", "E"]
)

example_series
````

````python
example_series.iloc[1:4]
````

Positions `1`, `2`, and `3` are returned.

## 7.2 Series: Label-Based Slicing

For `.loc[]`, the ending label is included.

````python
example_series.loc["B":"D"]
````

Labels `"B"`, `"C"`, and `"D"` are returned.

## 7.3 DataFrame: Position-Based Slicing with `.iloc[]`

The syntax is:

```python
df.iloc[row_start:row_stop, column_start:column_stop]
```

Both stop positions are excluded.

````python
students.iloc[1:4, 0:3]
````

This selects:

- row positions `1`, `2`, and `3`
- column positions `0`, `1`, and `2`

## 7.4 Creating Named Row Indices

The current `students` DataFrame uses default row labels.

Use `.set_index()` to make a column become the row index.

````python
students_by_name = students.set_index("Name")
students_by_name
````

Now `"Alice"`, `"Bob"`, `"Charlie"`, `"David"`, and `"Emma"` are row labels.

## 7.5 DataFrame: Label-Based Slicing with `.loc[]`

The syntax is:

```python
df.loc[row_start:row_stop, column_start:column_stop]
```

The ending labels are included.

````python
students_by_name.loc["Bob":"David"]
````

````python
students_by_name.loc["Bob":"David", "Major":"Score"]
````

## 7.6 Named Index Does Not Remove Integer Positions

Even after assigning named row labels, `.iloc[]` still accesses rows and columns by numbered position.

````python
students_by_name.iloc[1:4, 0:2]
````

<h2 style="margin-top:0;">✅ Key Takeaways</h2>

### Pandas Series
- A **Series** is a one-dimensional labeled collection.
- Every element has an integer **position**.
- Every element also has an **index label**.
- Index labels may contain **duplicates**; integer positions remain distinct.
- Use `.loc[]` to select by **label**.
- Use `.iloc[]` to select by **position**.
- Use `.str` to apply vectorized string operations.

### Pandas DataFrames
- A **DataFrame** is a two-dimensional table.
- A DataFrame can be created from a **dictionary**.
- `pd.read_csv()` reads a CSV file into a DataFrame.
- Each DataFrame **column is a Series**.
- Selecting one row normally also returns a Series.
- `head()` and `tail()` quickly preview rows.
- `drop()` removes columns.
- `rename()` changes column labels.
- `sort_values()` sorts rows by data values.
- `set_index()` creates named row labels.

### The Most Important Selection Rule
- **`.loc[]` → labels**
- **`.iloc[]` → integer positions**

### The Most Important Slicing Rule
- **`.iloc[]` stop position is excluded**
- **`.loc[]` ending label is included**
