```python
import jupyturtle
```

```python
alex = jupyturtle.Turtle()
```

```python
alex.drawing.bgcolor = 'blue'
alex.jump_to(alex.drawing.width/2, alex.drawing.height/2)
alex.draw()
```

```python
for i in ['yellow', 'red', 'orange']:
    alex.pen_color = i
    alex.pen_width = 10
    alex.forward(50)
    alex.left(120)
```
